```python
from bs4 import BeautifulSoup as bs 
import requests
```

```python
source_url = "https://en.wikipedia.org/wiki/Toy_Story"
r = requests.get(source_url)

soup = bs(r.content)

contents = soup.prettify()
print(contents)
```
